```python

import numpy as np

# Create a one-dimensional array
array = np.random.randint(1, 100, 9)
print("Original array:")
print(array)

# Square root of each element
print("\nSquare root:")
print(np.sqrt(array))

# Number of dimensions
print("\nOriginal array dimensions:")
print(array.ndim)

# Reshape into 3 rows and 3 columns
new_array = array.reshape(3, 3)
print("\nReshaped array:")
print(new_array)

# Number of dimensions after reshaping
print("\nReshaped array dimensions:")
print(new_array.ndim)

# Flatten the array into one dimension
print("\nFlattened array:")
print(new_array.ravel())

# Reshape again into 3x3
newm = new_array.reshape(3, 3)
print("\nNew array:")
print(newm)

# Select the third row, columns 2 and 3
print("\nnewm[2,1:3]:")
print(newm[2, 1:3])

# Select row 2, columns 2 and 3
print("\nnewm[1:2,1:3]:")
print(newm[1:2, 1:3])

# Select all rows and no columns
print("\nnew_array[0:3,0:0]:")
print(new_array[0:3, 0:0])

# Select the first two rows, first column
print("\nnew_array[0:2,0:1]:")
print(new_array[0:2, 0:1])

# Select all rows, first column
print("\nnew_array[0:3,0:1]:")
print(new_array[0:3, 0:1])

# Select rows 2 and 3
print("\nnew_array[1:3]:")
print(new_array[1:3])
```
