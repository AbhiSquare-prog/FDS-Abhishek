```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline

# Step 1: Load the Iris dataset
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)

# Rename columns to match your screenshot
data = data.rename(columns={
    'sepal_length': 'SepalLengthCm',
    'sepal_width': 'SepalWidthCm',
    'petal_length': 'PetalLengthCm',
    'petal_width': 'PetalWidthCm',
    'species': 'variety'
})

# Add ID column
data.insert(0, 'Id', range(1, len(data) + 1))

# Step 2: Display dataset
display(data)

# Step 3: Dataset information
data.info()

# Step 4: Statistical description
display(data.describe())

# Step 5: Count each variety
print(data['variety'].value_counts())

# Step 6: Count plot
sns.countplot(x='variety', data=data)
plt.title('Count of Each Iris Variety')
plt.show()

# Step 7: One-hot encoding
dummies = pd.get_dummies(data['variety'])

FinalDataset = pd.concat(
    [dummies, data.drop(columns=['variety'])],
    axis=1
)

display(FinalDataset.head())

# Step 8: Sepal length vs sepal width
sns.scatterplot(
    x='SepalLengthCm',
    y='SepalWidthCm',
    hue='variety',
    data=data
)
plt.title('Sepal Length vs Sepal Width')
plt.show()

# Step 9: Petal length vs petal width
sns.scatterplot(
    x='PetalLengthCm',
    y='PetalWidthCm',
    hue='variety',
    data=data
)
plt.title('Petal Length vs Petal Width')
plt.show()

# Step 10: Pair plot
sns.pairplot(data, hue='variety', height=3)
plt.show()

# Step 11: Petal length histogram
sns.FacetGrid(data, hue='variety', height=5).map(
    sns.histplot, 'PetalLengthCm'
).add_legend()
plt.show()

# Step 12: Petal width histogram
sns.FacetGrid(data, hue='variety', height=5).map(
    sns.histplot, 'PetalWidthCm'
).add_legend()
plt.show()

# Step 13: Sepal length histogram
sns.FacetGrid(data, hue='variety', height=5).map(
    sns.histplot, 'SepalLengthCm'
).add_legend()
plt.show()

# Step 14: Sepal width histogram
sns.FacetGrid(data, hue='variety', height=5).map(
    sns.histplot, 'SepalWidthCm'
).add_legend()
plt.show()
```
