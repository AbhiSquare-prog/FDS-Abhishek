```python

import numpy as np
import pandas as pd

# -----------------------------------
# PART 1: Create a simple DataFrame
# -----------------------------------

list1 = [[1, 'Smith', 50000], [2, 'Jones', 60000]]

df = pd.DataFrame(list1)
display(df)

# Rename columns
df.columns = ['EmpId', 'Name', 'Salary']
display(df)

# Display DataFrame information
df.info()

# -----------------------------------
# PART 2: Load the 50 Startups dataset
# -----------------------------------

url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/50_Startups.csv"

try:
    startup_df = pd.read_csv(url)
    print("\n50 Startups Dataset Information:")
    startup_df.info()

    print("\nFirst five rows:")
    display(startup_df.head())

    print("\nLast five rows:")
    display(startup_df.tail())

except Exception as e:
    print("Could not load the startup dataset:", e)

# -----------------------------------
# PART 3: Create the employee dataset
# -----------------------------------

employee_data = [
    [1, 'SREE VARSSINI K S', 5000],
    [2, 'SREEMATHI B', 6000],
    [3, 'SREYA G', 7000],
    [4, 'SREYASKARI MULLAPUDI', 5000],
    [5, 'SRI AKASH U G', 8000],
    [6, 'SRI HARSHAVARDHANAN R', 3000],
    [7, 'SRI HARSHAVARDHANAN R', 6000]
]

empCol = ['emp_id', 'name', 'salary']

df = pd.DataFrame(employee_data, columns=empCol)

# Display employee data
print("\nFirst five employees:")
display(df.head())

print("\nLast five employees:")
display(df.tail())

print("\nDataFrame information:")
df.info()

# Select salary column
print("\nSalary column:")
display(df.salary)

print("\nType of salary:")
print(type(df.salary))

# -----------------------------------
# PART 4: Statistical calculations
# -----------------------------------

print("\nMean salary:")
print(df.salary.mean())

print("\nMedian salary:")
print(df.salary.median())

print("\nMode salary:")
display(df.salary.mode())

print("\nSalary variance:")
print(df.salary.var())

print("\nSalary standard deviation:")
print(df.salary.std())

print("\nNumerical description:")
display(df.describe())

print("\nDescription including text columns:")
display(df.describe(include='all'))

# -----------------------------------
# PART 5: DataFrame columns and values
# -----------------------------------

print("\nEmployee columns:")
print(df.columns)

print("\nEmployee array:")
emparray = df.values
print(emparray)

# Recreate DataFrame from array
employee_DF = pd.DataFrame(emparray, columns=empCol)

print("\nRecreated employee DataFrame:")
display(employee_DF)
```
